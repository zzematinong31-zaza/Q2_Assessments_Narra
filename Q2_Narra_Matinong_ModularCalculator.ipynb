# ----------Modular Calculator----------

```python
# Asks the user for the input.
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

# Asks for the operation
operation = int(input("Choose operation: 1 - Addition 2 - Subtraction 3 - Division 4 - Multiplication:  "))

# Functions for the operations.
def addition(num1, num2):
  sum = num1 + num2
  return sum

def subtraction(num1, num2):
  difference = num2 - num1
  return difference

# Used a code where if the number is zero, it asks the user for another number.
def division(num1, num2):
  if num1 == 0:
    num1 = int(input("Number must not be zero: "))
    if num2 == 0:
     num2 = int(input("Number must not be zero: "))
  quotient = num1 / num2
  return quotient

def multiplication(num1, num2):
  product = num1 * num2
  return product

# Calls the operation depending on the users choice
if operation == 1:
  print(addition(num1, num2))
elif operation == 2:
  print(subtraction(num1, num2))
elif operation == 3:
   print(division(num1, num2))
else:
  print(multiplication(num1, num2))
```
